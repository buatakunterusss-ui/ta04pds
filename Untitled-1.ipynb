```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats

np.random.seed(301240019)   
```

```python
df = pd.read_csv("digital_marketing_campaign_dataset.csv")
print(df.shape)
df.info()
df.describe()
df.head()
```

```python
# Cek kualitas data
print("Missing value:", df.isnull().sum().sum())
print("Duplikat      :", df.duplicated().sum())

# Hapus outlier ekstrem (3 x IQR) pada ConversionRate
q1, q3 = df["ConversionRate"].quantile([0.25, 0.75])
iqr = q3 - q1
batas_bawah, batas_atas = q1 - 3*iqr, q3 + 3*iqr

df_clean = df[(df["ConversionRate"] >= batas_bawah) &
              (df["ConversionRate"] <= batas_atas)].copy()
print("Sebelum:", df.shape[0], "| Sesudah:", df_clean.shape[0])
```

```python
# --- ASUMSI (jelaskan di laporan) ---
PENGUNJUNG_PER_HARI = 1000   # asumsi jumlah pengunjung kampanye per hari
NILAI_PER_KONVERSI  = 50     # asumsi pendapatan per konversi (misal USD)
HARI                = 30

# --- PARAMETER DARI DATA ---
mu_cr    = df_clean["ConversionRate"].mean()
sigma_cr = df_clean["ConversionRate"].std()
print(f"Mean ConversionRate (μ)  : {mu_cr:.5f}")
print(f"Std Dev (σ)              : {sigma_cr:.5f}")

# Pendapatan harian ekspektasi (baseline)
mu_harian = PENGUNJUNG_PER_HARI * mu_cr * NILAI_PER_KONVERSI
print(f"Ekspektasi pendapatan/hari: {mu_harian:,.2f}")
```

```python
plt.figure(figsize=(8,5))
plt.hist(df_clean["ConversionRate"], bins=40, density=True,
         alpha=0.6, edgecolor="black", label="Data asli")
x = np.linspace(df_clean["ConversionRate"].min(), df_clean["ConversionRate"].max(), 200)
plt.plot(x, stats.norm.pdf(x, mu_cr, sigma_cr), "r-", label="Kurva Normal")
plt.title("Distribusi ConversionRate (Data Asli)")
plt.xlabel("ConversionRate"); plt.ylabel("Densitas"); plt.legend()
plt.show()

print("Skewness:", df_clean["ConversionRate"].skew())
print(stats.shapiro(df_clean["ConversionRate"].sample(500, random_state=1)))
```

```python
baseline_total = mu_harian * HARI        # ekspektasi 30 hari
threshold = baseline_total * 0.90        # turun lebih dari 10%
print(f"Baseline : {baseline_total:,.2f}")
print(f"Threshold: {threshold:,.2f}")
```

```python
n_sim = 10_000

# Tiap baris = 1 skenario, tiap kolom = 1 hari
cr_acak = np.random.normal(mu_cr, sigma_cr, size=(n_sim, HARI))
cr_acak = np.clip(cr_acak, 0, None)      # conversion rate tidak boleh negatif

pendapatan_harian = PENGUNJUNG_PER_HARI * cr_acak * NILAI_PER_KONVERSI
total_revenue = pendapatan_harian.sum(axis=1)   # total 30 hari per skenario
```

```python
prob_turun = np.sum(total_revenue < threshold) / n_sim
print(f"P(pendapatan turun > 10%) = {prob_turun:.4f} ({prob_turun*100:.2f}%)")
print(f"Rata-rata simulasi : {total_revenue.mean():.2f}")
print(f"Persentil 5% (VaR) : {np.percentile(total_revenue, 5):.2f}")
```

```python
plt.figure(figsize=(9,5))
plt.hist(total_revenue, bins=60, alpha=0.7, edgecolor="black")
plt.axvline(threshold, color="red", linestyle="--", label=f"Threshold (-10%) = {threshold:,.0f}")
plt.axvline(baseline_total, color="green", linestyle="-", label="Baseline")
plt.title(f"Simulasi Monte Carlo ({n_sim:,} skenario): P(turun>10%) = {prob_turun:.2%}")
plt.xlabel("Total Pendapatan 30 Hari"); plt.ylabel("Frekuensi"); plt.legend()
plt.show()
```

```python
for pct in [0.05, 0.10, 0.15, 0.20]:
    p = np.mean(total_revenue < baseline_total * (1 - pct))
    print(f"Turun > {pct:.0%}: {p:.4f}")
```
